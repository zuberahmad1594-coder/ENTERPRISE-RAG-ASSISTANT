# 01 — Document Exploration

Explore the raw HR document collection: how many documents, how many pages, what does the extracted text look like, and are there any extraction quality issues to address before chunking.

```python
import sys, os
sys.path.insert(0, os.path.join(os.getcwd(), '..', 'src'))

from document_loader import load_documents

DOCS_DIR = os.path.join('..', 'data', 'documents')
records = load_documents(DOCS_DIR)
print(f'Loaded {len(records)} page records')
```

```python
from collections import Counter
doc_page_counts = Counter(r.doc_name for r in records)
for doc, count in doc_page_counts.items():
    print(f'{doc}: {count} pages/sections extracted')
```

```python
# Inspect a sample of extracted, cleaned text
for r in records[:2]:
    print(f'--- {r.doc_name} | page {r.page_number} ---')
    print(r.text)
    print()
```

```python
# Basic text stats — useful for deciding chunk size in the next notebook
word_counts = [len(r.text.split()) for r in records]
print('Min words/page:', min(word_counts))
print('Max words/page:', max(word_counts))
print('Avg words/page:', sum(word_counts) / len(word_counts))
```
